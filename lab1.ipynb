# Разминка: решаем задачи

В финале игры «Звёздный час» игрокам нужно выписать как можно больше слов, составленных из букв исходного слова. Для составления каждого слова разрешается использовать часть или даже все буквы исходного, но нельзя ничего добавлять: ни дополнительные копии имеющихся в слове букв, ни, тем более, другие буквы. Проанализируйте список слов, получившийся у одного из игроков: подсчитайте, сколько слов составлено в соответствии с правилами, и выведите эти слова.

## Формат ввода
Первым идёт исходное слово. Затем слова анализируемого списка. Каждое слово записано на отдельной строке.

## Формат вывода
На первой строке запишите одно число — количество слов, составленных по правилам (мы проверяем только буквенный состав каждого слова, всё остальное не в нашей юрисдикции, даже повторы слов). В каждой следующей строке выводите по одному такому слову в том же порядке, в каком они встретились в исходном списке.

## Пример

### Ввод: 
лесоповал

лес

повар

вопль

сопло

весло

село

слава

### Вывод:
4

лес

сопло

весло

село

## Задача 2

К празднику в детском саду решили поставить спектакль «Репка». Точный список персонажей и порядок их выхода на сцену пока не утверждены и периодически меняются, поэтому начинающим актёрам требуются подсказки, после кого им выходить и кто идёт после них.

### Формат ввода
В первой строке задаётся актуальный на данный момент порядок выхода персонажей на сцену в виде перечисления их имён через стрелочку ->. Во второй строке записывается количество запросов. Каждая следующая строка представляет собой имя персонажа, которому требуется подсказка. Все запросы корректны.

### Формат вывода
На каждый запрос необходимо ответить на отдельной строке. Ответ персонажу должен содержать имя того, после кого ему выходить, его собственное имя и имя следующего за ним. Имена разделяются стрелочками, как и во входных данных. Если персонаж выходит первым или последним, для него указывается только один сосед.

### Пример
Репка -> Дед -> Бабка -> Внучка -> Жучка -> Кошка -> Мышка

3

Репка

Мышка

Бабка
#### Ответ
Репка -> Дед

Кошка -> Мышка

Дед -> Бабка -> Внучка

Напишем часть логики для робота – продавца мороженого. Сначала загружается меню: количество имеющихся сортов мороженого и перечень их наименований с ценами. # Шунтирующий двор → ОПЗ

## Обратная польская нотация

### Пример 1: Простое арифметическое выражение
- **Инфиксная запись**: `3 + 4 * 2`
- **ОПН**: `3 4 2 * +`
- **Результат**: 11

### Пример 2: Выражение со скобками
- **Инфиксная запись**: `(5 + 3) * 2 - 1`
- **ОПН**: `5 3 + 2 * 1 -`
- **Результат**: 15

### Пример 3: Сложное выражение
- **Инфиксная запись**: `2 * (3 + 4) / (5 - 2)`
- **ОПН**: `2 3 4 + * 5 2 - /`
- **Результат**: 4.67

**Преимущества ОПН**: отсутствие скобок, однозначность вычислений, простота реализации в стековых вычислителях.

````python
class Stack:
    '''
    переменные класса = атрибуты
    функции класса = методы
    '''
    #new_items = ''
    def __init__(self, *args): #__new__
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        return self.items.pop()

    def is_empty(self):
        return (self.items == [])
    
stack = Stack()
stack.push(5)
print(stack.pop())
print(stack.is_empty())
````

# Дерево исключений

https://tatyderb.gitbooks.io/python-express-course/content/chapter_exception/3_tree.html

````python
class MyCustomError(Exception):
    pass

class ValidationError(Exception):
    def __init__(self, message, field=None):
        super().__init__(message)
        self.field = field

class DatabaseError(Exception):
    def __init__(self, message, error_code=None):
        super().__init__(message)
        self.error_code = error_code

# Простое исключение
raise MyCustomError("Что-то пошло не так")

# С дополнительными атрибутами
raise ValidationError("Неверный формат email", field="email")
raise DatabaseError("Соединение потеряно", error_code=500)

# Обработка
try:
    raise ValidationError("Ошибка валидации", field="username")
except ValidationError as e:
    print(f"Ошибка в поле {e.field}: {e}")
````

# Фикстуры в pytest

## Определение
Фикстура — это функция, которая подготавливает данные или ресурсы для тестов и выполняет очистку после их завершения.

## Основные задачи фикстур
- Подготовка тестовых данных
- Создание объектов для тестирования
- Настройка окружения
- Очистка ресурсов после тестов

## Простой пример

```python
import pytest

@pytest.fixture
def sample_data():
    return [1, 2, 3, 4, 5]

def test_sum(sample_data):
    assert sum(sample_data) == 15
```

## Именование файлов и функций

### Правила именования:
- Файлы: `test_*.py` или `*_test.py`
- Функции: `test_*()`
- Классы: `Test*`
- Методы в классах: `test_*()`

```python
# test_calculator.py
class TestCalculator:
    def test_add(self):
        assert 2 + 2 == 4
    
    def test_subtract(self):
        assert 5 - 3 == 2

def test_multiply():
    assert 3 * 4 == 12
```

## Конфигурация pytest.ini

```ini
[tool:pytest]
testpaths = tests
python_files = test_*.py *_test.py
python_classes = Test*
python_functions = test_*
addopts = -v --tb=short
markers =
    slow: marks tests as slow
    integration: marks tests as integration tests
```

## Способы запуска

### Базовые команды
```bash
# Все тесты
pytest

# Конкретный файл
pytest tests/test_calculator.py

# Конкретный тест
pytest tests/test_calculator.py::test_add

# Конкретный метод класса
pytest tests/test_calculator.py::TestCalculator::test_add

# По шаблону
pytest tests/ -k "test_add"
```

### Полезные опции
```bash
# Подробный вывод
pytest -v

# Очень подробный
pytest -vv

# Остановиться на первой ошибке
pytest -x

# Остановиться после N ошибок
pytest --maxfail=3

# Показать локальные переменные при ошибках
pytest -l

# Запустить только неудачные тесты из последнего запуска
pytest --lf

# Запустить сначала неудачные, потом остальные
pytest --ff
```
